```python
from matplotlib.colors import ListedColormap
def plot_decision_regions(X, y, classifier, resolution=0.02):

    # setup marker generator and color map
    markers = ('s', 'x', 'o', '^', 'v')
    colors = ('red', 'blue', 'lightgreen', 'gray', 'cyan')
    cmap = ListedColormap(colors[:len(np.unique(y))])

    # plot the decision surface
    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, resolution),
                           np.arange(x2_min, x2_max, resolution))
    Z = classifier.predict(np.array([xx1.ravel(), xx2.ravel()]).T)
    Z = (Z > 0.5).astype(int)  # predict devuelve probabilidades: > 0.5 es clase 1
    Z = Z.reshape(xx1.shape)
    plt.contourf(xx1, xx2, Z, alpha=0.3, cmap=cmap)
    plt.xlim(xx1.min(), xx1.max())
    plt.ylim(xx2.min(), xx2.max())

    # plot class samples
    for idx, cl in enumerate(np.unique(y)):
        plt.scatter(x=X[y == cl, 0],
                    y=X[y == cl, 1],
                    alpha=0.8,
                    c=colors[idx],
                    marker=markers[idx],
                    label=cl,
                    edgecolor='black')
```

14. ¿Sería posible representar gráficamente los límites de decisión con 4 variables de entrada? ¿Y si utilizáramos 3 variables únicamente? Entrenamiento con 3 variables y límites de decisión en 3D.

```python
# 3 variables: longitud de sépalo, longitud de pétalo y anchura de pétalo
X3 = df.iloc[0:150, [0, 2, 3]].values
y3 = df.iloc[0:150, 4].values
y3 = np.where(y3 == 'Iris-setosa', 0, np.where(y3 == 'Iris-versicolor', 1, 2))
X3, y3 = shuffle(X3, y3, random_state=42)

X3_std = np.copy(X3)
X3_std[:, 0] = (X3[:, 0] - X3[:, 0].mean()) / X3[:, 0].std()
X3_std[:, 1] = (X3[:, 1] - X3[:, 1].mean()) / X3[:, 1].std()
X3_std[:, 2] = (X3[:, 2] - X3[:, 2].mean()) / X3[:, 2].std()

tf.random.set_seed(42)
model5 = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(3,), name='input'),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
model5.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01), loss=tf.keras.losses.sparse_categorical_crossentropy, metrics=['accuracy'])
hist5 = model5.fit(X3_std, y3, epochs=200, verbose=0)
print('accuracy:', hist5.history['accuracy'][-1])
```

```python
# Rejilla de puntos en 3D: cada punto se pinta del color de la clase que predice la RNA
valores = np.linspace(-2.5, 2.5, 11)
g1, g2, g3 = np.meshgrid(valores, valores, valores)
rejilla = np.array([g1.ravel(), g2.ravel(), g3.ravel()]).T
clase = np.argmax(model5.predict(rejilla, verbose=0), axis=1)

colores = np.array(['red', 'blue', 'green'])
fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(projection='3d')
ax.scatter(X3_std[:, 0], X3_std[:, 1], X3_std[:, 2], c=colores[y3], s=25, edgecolor='black')
ax.set_xlabel('long. sépalo')
ax.set_ylabel('long. pétalo')
ax.set_zlabel('anch. pétalo')
plt.show()
```
